# Problem 3: Checking Routes Between Cities Using DFS

# A road network connects several cities.

# Tasks:
# 1. Check whether a route exists between two given cities.
# 2. Find all connected components of the road network.

# We use DFS to explore the graph.

```python
cities = {
    'Delhi': ['Mumbai', 'Jaipur'],
    'Mumbai': ['Delhi', 'Pune'],
    'Jaipur': ['Delhi'],
    'Pune': ['Mumbai'],

    'Kolkata': ['Patna'],
    'Patna': ['Kolkata'],

    'Chennai': []
}

print("Road Network:")

for city, connections in cities.items():
    print(city, "->", connections)
```

```python
def route_exists(graph, start, destination, visited=None):

    if visited is None:
        visited = set()

    # We reached the destination
    if start == destination:
        return True

    visited.add(start)

    for neighbor in graph[start]:

        if neighbor not in visited:

            if route_exists(graph, neighbor, destination, visited):
                return True

    return False
```

```python
# Test 1
start = 'Delhi'
destination = 'Pune'

result = route_exists(cities, start, destination)

print(f"Route from {start} to {destination}: {result}")


# Test 2
start = 'Delhi'
destination = 'Patna'

result = route_exists(cities, start, destination)

print(f"Route from {start} to {destination}: {result}")
```

```python
def dfs_component(graph, city, visited, component):

    visited.add(city)
    component.append(city)

    for neighbor in graph[city]:

        if neighbor not in visited:
            dfs_component(
                graph,
                neighbor,
                visited,
                component
            )


def find_connected_components(graph):

    visited = set()
    components = []

    for city in graph:

        if city not in visited:

            component = []

            dfs_component(
                graph,
                city,
                visited,
                component
            )

            components.append(component)

    return components
```

```python
components = find_connected_components(cities)

print("Connected Components:\n")

for i, component in enumerate(components, start=1):

    print(f"Component {i}: {component}")
```
