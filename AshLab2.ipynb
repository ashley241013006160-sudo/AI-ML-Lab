# Problem 2: The Water Jug Problem

# Jug A has a capacity of 4 litres.
# Jug B has a capacity of 3 litres.

# Allowed operations:
# 1. Fill a jug completely.
# 2. Empty a jug completely.
# 3. Pour water from one jug into the other.

# Goal:
# Measure exactly 2 litres of water in either jug.

# We use BFS because BFS guarantees the minimum number of steps.

```python
from collections import deque

def water_jug_bfs(capacity_a, capacity_b, target):

    # State = (amount in Jug A, amount in Jug B)
    start = (0, 0)

    queue = deque([start])

    visited = {start}

    parent = {start: None}

    action = {start: "Start"}

    while queue:

        current = queue.popleft()

        a, b = current

        # Check whether target has been reached
        if a == target or b == target:

            path = []

            while current is not None:
                path.append((current, action[current]))
                current = parent[current]

            path.reverse()

            return path

        next_states = []

        # 1. Fill Jug A
        next_states.append(
            ((capacity_a, b), "Fill Jug A")
        )

        # 2. Fill Jug B
        next_states.append(
            ((a, capacity_b), "Fill Jug B")
        )

        # 3. Empty Jug A
        next_states.append(
            ((0, b), "Empty Jug A")
        )

        # 4. Empty Jug B
        next_states.append(
            ((a, 0), "Empty Jug B")
        )

        # 5. Pour A -> B
        amount = min(a, capacity_b - b)

        next_states.append(
            (
                (a - amount, b + amount),
                "Pour Jug A -> Jug B"
            )
        )

        # 6. Pour B -> A
        amount = min(b, capacity_a - a)

        next_states.append(
            (
                (a + amount, b - amount),
                "Pour Jug B -> Jug A"
            )
        )

        for next_state, operation in next_states:

            if next_state not in visited:

                visited.add(next_state)

                parent[next_state] = current
                action[next_state] = operation

                queue.append(next_state)

    return None


solution = water_jug_bfs(4, 3, 2)

print("Minimum sequence of steps:\n")

for i, (state, operation) in enumerate(solution):

    print(f"Step {i}: {operation}")
    print(f"       Jug A = {state[0]} L, Jug B = {state[1]} L")
```
